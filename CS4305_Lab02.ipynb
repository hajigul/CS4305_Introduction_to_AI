## Demo 1: Machine Learning — Decision Tree
**Aim:** Let the computer *learn* rules to identify iris flower species from 4 measurements.

**Steps:** load data → split 70% train / 30% test → train → test on unseen data → predict a new flower.

````python
# ===== Demo 1: Machine Learning - Decision Tree Classifier =====
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# 1. Load the data (Experience E)
iris = load_iris()
X, y = iris.data, iris.target          # X = measurements, y = species (0, 1, 2)
print("Features:", iris.feature_names)
print("Classes :", list(iris.target_names))

# 2. Split: 70% for learning, 30% for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Train the model (the computer learns the rules itself)
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)

# 4. Test on data the model has never seen (Performance measure P)
y_pred = model.predict(X_test)
print("Accuracy on unseen test data:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")

# 5. Predict a brand-new flower
new_flower = [[5.1, 3.5, 1.4, 0.2]]    # sepal length, sepal width, petal length, petal width (cm)
print("Predicted species:", iris.target_names[model.predict(new_flower)[0]])

# Visualize the rules the computer learned
plt.figure(figsize=(14, 7))
plot_tree(model, feature_names=iris.feature_names, class_names=iris.target_names,
          filled=True, rounded=True)
plt.title("Rules learned automatically from data")
plt.show()
````

**Observe:** nobody wrote the IF-THEN rules in the tree diagram — the model discovered them from data.

## Demo 2: Neural Network — Perceptron Learning the AND Gate
**Aim:** Train a single artificial neuron. It starts with zero weights and corrects itself after every mistake:

`new weight = old weight + learning_rate × error × input`

````python
# ===== Demo 2: Perceptron learning the AND gate (from scratch with NumPy) =====
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])   # inputs
y = np.array([0, 0, 0, 1])                        # correct AND outputs

w = np.zeros(2)     # weights start at zero - the neuron knows nothing
b = 0.0             # bias
lr = 0.1            # learning rate

def step(z):
    return 1 if z >= 0 else 0

for epoch in range(1, 21):
    errors = 0
    for xi, target in zip(X, y):
        output = step(np.dot(w, xi) + b)
        error = target - output
        if error != 0:                 # learn only from mistakes
            w += lr * error * xi
            b += lr * error
            errors += 1
    print(f"Epoch {epoch}: weights={w}, bias={b:.1f}, errors={errors}")
    if errors == 0:
        print("Training complete - the perceptron has learned AND!")
        break

print("\nTesting the trained perceptron:")
for xi in X:
    print(f"{xi[0]} AND {xi[1]} = {step(np.dot(w, xi) + b)}")
````

**Try:** change `y` to `[0, 1, 1, 1]` (OR) and run again. Then try XOR `[0, 1, 1, 0]` — it never reaches 0 errors, because one perceptron cannot learn XOR.

## Demo 3: Natural Language Processing with NLTK
**Aim:** Run a basic NLP pipeline: sentence tokenization → word tokenization → stop-word removal → stemming.

````python
# ===== Demo 3: NLP basics with NLTK =====
import nltk
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)     # needed by newer NLTK versions in Colab
nltk.download("stopwords", quiet=True)

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

text = ("Artificial Intelligence is changing the world. Students are learning AI "
        "in universities across Pakistan. Learning AI is exciting and very useful!")

# Step 1: Sentence tokenization
sentences = sent_tokenize(text)
print("Sentences:", sentences)

# Step 2: Word tokenization
words = word_tokenize(text.lower())
print("\nTokens:", words)

# Step 3: Remove stop words (common words like 'is', 'the', 'and') and punctuation
stop_words = set(stopwords.words("english"))
content_words = [w for w in words if w.isalpha() and w not in stop_words]
print("\nAfter stop-word removal:", content_words)

# Step 4: Stemming (reduce words to their root form)
stemmer = PorterStemmer()
print("\nStems:", [stemmer.stem(w) for w in content_words])
````

### Simple rule-based sentiment analysis
Uses positive/negative word lists (knowledge), not learning.

````python
# ===== Simple lexicon-based sentiment analysis =====
positive = {"good", "great", "excellent", "love", "amazing", "helpful", "exciting", "useful", "best"}
negative = {"bad", "poor", "terrible", "hate", "boring", "useless", "worst", "slow", "difficult"}

def simple_sentiment(sentence):
    tokens = word_tokenize(sentence.lower())
    score = sum(t in positive for t in tokens) - sum(t in negative for t in tokens)
    label = "Positive" if score > 0 else "Negative" if score < 0 else "Neutral"
    return label, score

reviews = ["The AI course is excellent and very helpful.",
           "The lab was boring and the computers were slow.",
           "The class starts at 9 AM."]
for r in reviews:
    print(f"{r:50s} -> {simple_sentiment(r)}")
````

**Try:** `simple_sentiment("The course is not good.")` — it says *Positive*. Why? It doesn't understand negation.

## Demo 4: Computer Vision — Edge Detection with OpenCV
**Aim:** See how a computer treats an image as numbers and extracts edges (a basic visual feature).

*Note:* `cv2.imshow()` does not work in Colab, so we display images with Matplotlib.

````python
# ===== Demo 4: Computer Vision - Edge detection =====
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data            # built-in sample images (no upload needed)

img = data.coffee()                              # an RGB image of a coffee cup
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)     # step 1: convert to grayscale
blur = cv2.GaussianBlur(gray, (5, 5), 0)         # step 2: remove noise
edges = cv2.Canny(blur, 50, 150)                 # step 3: detect edges

print("Image shape (rows, cols, channels):", img.shape)
print("Top-left pixel (R, G, B):", img[0, 0])
print("Edge pixels found:", np.count_nonzero(edges), "out of", edges.size)

# NOTE: cv2.imshow() does not work in Colab, so we use matplotlib
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
titles = ["1. Original (what the camera sees)", "2. Grayscale (pixel intensities)",
          "3. Edges detected (Canny)"]
for ax, image, title in zip(axes, [img, gray, edges], titles):
    ax.imshow(image, cmap=None if image.ndim == 3 else "gray")
    ax.set_title(title)
    ax.axis("off")
plt.tight_layout()
plt.show()
````

## Demo 5: Mini Medical Expert System
**Aim:** Build the three parts of an expert system — **knowledge base** (rules), **inference engine** (matching), **explanation facility** ("because of…").

*For teaching only — not a real medical tool.*

````python
# ===== Demo 5: Mini rule-based Expert System =====
# (For teaching only - not a real medical tool!)

# KNOWLEDGE BASE: rules obtained from an "expert"
knowledge_base = [
    {"disease": "Flu",                "symptoms": {"fever", "cough", "body_ache"}},
    {"disease": "Common Cold",        "symptoms": {"sneezing", "runny_nose", "sore_throat"}},
    {"disease": "Dengue (suspected)", "symptoms": {"fever", "headache", "joint_pain", "rash"}},
    {"disease": "Food Poisoning",     "symptoms": {"vomiting", "stomach_pain", "diarrhea"}},
]

# INFERENCE ENGINE: match the patient's facts against every rule
def diagnose(patient_symptoms):
    patient_symptoms = set(patient_symptoms)
    results = []
    for rule in knowledge_base:
        matched = rule["symptoms"] & patient_symptoms          # set intersection
        if matched:
            confidence = len(matched) / len(rule["symptoms"])
            results.append((rule["disease"], confidence, matched))
    return sorted(results, key=lambda r: r[1], reverse=True)

# EXPLANATION FACILITY: show WHY each conclusion was reached
def explain(patient_symptoms):
    print("Patient symptoms:", ", ".join(patient_symptoms))
    results = diagnose(patient_symptoms)
    if not results:
        print("  No rule matched. Please consult a doctor.\n")
        return
    for disease, conf, matched in results:
        print(f"  -> {disease}: {conf:.0%} match (because of: {', '.join(sorted(matched))})")
    print()

explain(["fever", "cough", "body_ache"])
explain(["fever", "headache", "joint_pain"])
explain(["sneezing", "stomach_pain"])
explain(["back_pain"])
````

**Observe:** the last patient matches no rule — expert systems are *brittle* outside their knowledge.

## Demo 6: Fuzzy Logic — Fan-Speed Controller
**Aim:** Represent Cold / Warm / Hot as fuzzy sets (degrees between 0 and 1) and compute a smooth fan speed.

````python
# ===== Demo 6: Fuzzy Logic - temperature membership and fan speed =====
import numpy as np
import matplotlib.pyplot as plt

def triangular(x, a, b, c):
    """Triangular membership function: 0 at a, rises to 1 at b, falls to 0 at c."""
    return np.maximum(np.minimum((x - a) / (b - a + 1e-9), (c - x) / (c - b + 1e-9)), 0)

temp = np.linspace(0, 50, 501)
cold = triangular(temp, -1, 0, 20)
warm = triangular(temp, 10, 25, 40)
hot  = triangular(temp, 30, 50, 51)

plt.figure(figsize=(9, 4))
plt.plot(temp, cold, label="Cold", color="blue")
plt.plot(temp, warm, label="Warm", color="orange")
plt.plot(temp, hot,  label="Hot",  color="red")
plt.xlabel("Temperature (°C)"); plt.ylabel("Degree of membership")
plt.title("Fuzzy sets for temperature"); plt.legend(); plt.grid(alpha=0.3)
plt.show()

# Fuzzy rules:  IF cold THEN fan = 0%;  IF warm THEN fan = 50%;  IF hot THEN fan = 100%
def fan_speed(t):
    mu_cold = float(triangular(t, -1, 0, 20))
    mu_warm = float(triangular(t, 10, 25, 40))
    mu_hot  = float(triangular(t, 30, 50, 51))
    # Defuzzification: weighted average of rule outputs
    speed = (mu_cold * 0 + mu_warm * 50 + mu_hot * 100) / (mu_cold + mu_warm + mu_hot)
    print(f"Temp {t:2d}°C -> Cold={mu_cold:.2f}, Warm={mu_warm:.2f}, Hot={mu_hot:.2f}"
          f" -> Fan speed = {speed:.0f}%")

for t in [5, 15, 25, 35, 45]:
    fan_speed(t)
````

**Observe:** at 15 °C the temperature is partly cold (0.25) *and* partly warm (0.33), so the fan speed changes smoothly instead of jumping.
